# ============================================================================
# CELLULE 1 : Importation des bibliothèques
# ============================================================================

```python

"""
Cette cellule importe toutes les bibliothèques nécessaires :
- cv2 : OpenCV pour le traitement d'images
- requests : pour télécharger des images depuis internet
- os : pour gérer les fichiers et dossiers
- numpy : pour manipuler les arrays (inclus avec OpenCV)
"""
import cv2
import requests
import os
import numpy as np
```


# ============================================================================
# CELLULE 2 : Fonction pour télécharger une image
# ===================================================================

```python
"""
Cette fonction télécharge une image depuis une URL et la sauvegarde localement.

Paramètres:
- url : l'URL de l'image à télécharger
- nom_fichier : le nom sous lequel sauvegarder l'image

Retourne:
- chemin_fichier : le chemin complet du fichier téléchargé
"""

def telecharger_image(url, nom_fichier="animal.jpg"):
    try:
        # Créer un dossier 'images' s'il n'existe pas
        if not os.path.exists('images'):
            os.makedirs('images')
            print("✓ Dossier 'images' créé")
        
        # Chemin complet du fichier
        chemin_fichier = os.path.join('images', nom_fichier)
        
        # Télécharger l'image
        print(f" Téléchargement de l'image depuis: {url}")
        response = requests.get(url, timeout=10)
        response.raise_for_status()  # Vérifie que la requête a réussi
        
        # Sauvegarder l'image
        with open(chemin_fichier, 'wb') as f:
            f.write(response.content)
        
        print(f"✓ Image téléchargée avec succès : {chemin_fichier}")
        return chemin_fichier
    
    except requests.exceptions.RequestException as e:
        print(f"✗ Erreur lors du téléchargement : {e}")
        return None
    except Exception as e:
        print(f"✗ Erreur inattendue : {e}")
        return None
```

# ============================================================================
# CELLULE 3 : Télécharger une image d'un animal
# ============================================================================

```python
"""
Télécharge une image d'un chat depuis une URL publique.
Vous pouvez remplacer cette URL par n'importe quelle autre image.
"""

# URL d'exemple d'un chat (image libre de droits)
url_image = "https://images.unsplash.com/photo-1514888286974-6c03e2ca1dba?w=800"

# Télécharger l'image
chemin_image = telecharger_image(url_image, "chat.jpg")
```

# ============================================================================
# CELLULE 4 : Charger l'image avec OpenCV et vérifier
# ============================================================================

```python
"""
Charge l'image téléchargée avec OpenCV et vérifie si le chargement a réussi.
OpenCV retourne None si l'image ne peut pas être chargée.
"""

if chemin_image:
    # Charger l'image
    print(f"Chargement de l'image avec OpenCV...")
    image = cv2.imread(chemin_image)
    
    # Vérifier si l'image a été chargée correctement
    if image is not None:
        print("✓ Image chargée avec succès!")
        
        # Afficher les informations sur l'image
        hauteur, largeur, canaux = image.shape
        print(f"  → Dimensions : {largeur} x {hauteur} pixels")
        print(f"  → Nombre de canaux : {canaux} (BGR)")
        print(f"  → Type de données : {image.dtype}")
        print(f"  → Taille en mémoire : {image.nbytes} bytes")
    else:
        print("✗ Échec du chargement de l'image avec OpenCV")
        print("  Vérifiez que le fichier existe et est une image valide")
else:
    print("✗ Impossible de charger l'image car le téléchargement a échoué")
```

# ============================================================================
# CELLULE 5 : Redimensionner l'image (Resizing)
# ============================================================================

```python
"""
Redimensionne l'image à différentes tailles pour démontrer plusieurs approches.
OpenCV offre plusieurs méthodes d'interpolation pour le redimensionnement.
"""

if image is not None:
    print("\n" + "="*60)
    print("REDIMENSIONNEMENT DE L'IMAGE")
    print("="*60)
    
    # Méthode 1 : Redimensionner à une taille fixe
    nouvelle_largeur = 400
    nouvelle_hauteur = 300
    image_redim_1 = cv2.resize(image, (nouvelle_largeur, nouvelle_hauteur))
    print(f"✓ Image redimensionnée à {nouvelle_largeur}x{nouvelle_hauteur}")
    
    # Méthode 2 : Redimensionner avec un facteur d'échelle (50% de la taille originale)
    facteur_echelle = 0.5
    image_redim_2 = cv2.resize(image, None, fx=facteur_echelle, fy=facteur_echelle)
    nouvelle_dim = image_redim_2.shape
    print(f"✓ Image redimensionnée à 50% : {nouvelle_dim[1]}x{nouvelle_dim[0]}")
    
    
    # Sauvegarder les images redimensionnées
    cv2.imwrite('images/chat_redim_400x300.jpg', image_redim_1)
    cv2.imwrite('images/chat_redim_50pourcent.jpg', image_redim_2)

    print("\n✓ Toutes les images redimensionnées ont été sauvegardées dans le dossier 'images/'")
```

# ============================================================================
# CELLULE 6 : Affichage des informations finales
# ============================================================================

```python
"""
Résumé de toutes les opérations effectuées.
"""

if image is not None:
    print("\n" + "="*60)
    print("RÉSUMÉ")
    print("="*60)
    print(f"Image originale : {largeur}x{hauteur}")
    print(f"Image 1 (taille fixe) : {nouvelle_largeur}x{nouvelle_hauteur}")
    print(f"Image 2 (50% échelle) : {image_redim_2.shape[1]}x{image_redim_2.shape[0]}")
    print("\n✓ Traitement terminé avec succès!")
else:
    print("\n✗ Aucune image n'a pu être traitée")
```

# ============================================================================
# NOTES SUPPLÉMENTAIRES
# ============================================================================

# ============================================================================
# CELLULE 7 : Crop (Recadrage) de l'image
# ============================================================================

```python
"""
Le crop permet d'extraire une région spécifique de l'image.
En OpenCV, on utilise le slicing des arrays NumPy : image[y1:y2, x1:x2]
"""

if image is not None:
    print("\n" + "="*60)
    print("CROP (RECADRAGE)")
    print("="*60)
    
    # Définir la région à extraire (centre de l'image)
    x_debut = largeur // 4
    y_debut = hauteur // 4
    x_fin = (largeur * 3) // 4
    y_fin = (hauteur * 3) // 4
    
    # Effectuer le crop
    image_crop = image[y_debut:y_fin, x_debut:x_fin]
    
    print(f"✓ Image originale : {largeur}x{hauteur}")
    print(f"✓ Image croppée : {image_crop.shape[1]}x{image_crop.shape[0]}")
    print(f"  → Région extraite : du point ({x_debut},{y_debut}) au point ({x_fin},{y_fin})")
    
    # Sauvegarder
    cv2.imwrite('images/chat_crop.jpg', image_crop)
    print("✓ Image croppée sauvegardée")
```

# ============================================================================
# CELLULE 8 : Inversion de l'image
# ============================================================================

```python
"""
L'inversion inverse les couleurs de l'image (négatif).
Pour une image 8-bit, on fait : 255 - valeur_pixel
"""

if image is not None:
    print("\n" + "="*60)
    print("INVERSION (NÉGATIF)")
    print("="*60)
    
    # Méthode 1 : Inversion avec bitwise NOT
    image_inverse = cv2.bitwise_not(image)
    print("✓ Image inversée (négatif) créée avec bitwise_not()")
    
    # Méthode 2 : Inversion manuelle (alternative)
    image_inverse_alt = 255 - image
    print("✓ Image inversée (négatif) créée avec soustraction")
    
    # Sauvegarder
    cv2.imwrite('images/chat_inverse.jpg', image_inverse)
    print("✓ Image inversée sauvegardée")
```

# ============================================================================
# CELLULE 9 : Rotation de l'image
# ============================================================================

```python
"""
OpenCV offre plusieurs façons de faire une rotation :
1. Rotation simple à 90°, 180°, 270°
2. Rotation à n'importe quel angle avec getRotationMatrix2D()
"""

if image is not None:
    print("\n" + "="*60)
    print("ROTATION")
    print("="*60)
    
    # Rotation 1 : 90° dans le sens horaire
    image_rotation_90 = cv2.rotate(image, cv2.ROTATE_90_CLOCKWISE)
    print("✓ Rotation 90° horaire effectuée")
    
    # Rotation 2 : 180°
    image_rotation_180 = cv2.rotate(image, cv2.ROTATE_180)
    print("✓ Rotation 180° effectuée")
    
    # Rotation 3 : Angle personnalisé (45°)
    angle = 45
    centre = (largeur // 2, hauteur // 2)
    matrice_rotation = cv2.getRotationMatrix2D(centre, angle, 1.0)
    image_rotation_45 = cv2.warpAffine(image, matrice_rotation, (largeur, hauteur))
    print(f"✓ Rotation {angle}° effectuée")
    
    # Sauvegarder
    cv2.imwrite('images/chat_rotation_90.jpg', image_rotation_90)
    cv2.imwrite('images/chat_rotation_180.jpg', image_rotation_180)
    cv2.imwrite('images/chat_rotation_45.jpg', image_rotation_45)
    print("✓ Toutes les rotations sauvegardées")
```

# ============================================================================
# CELLULE 10 : Découpage (Slicing) avancé
# ============================================================================

```python
"""
Le découpage permet d'extraire plusieurs parties de l'image.
On peut créer une mosaïque ou extraire des régions spécifiques.
"""

if image is not None:
    print("\n" + "="*60)
    print("DÉCOUPAGE (SLICING)")
    print("="*60)
    
    # Découper l'image en 4 quadrants
    milieu_x = largeur // 2
    milieu_y = hauteur // 2
    
    quadrant_1 = image[0:milieu_y, 0:milieu_x]  # Haut-gauche
    quadrant_2 = image[0:milieu_y, milieu_x:largeur]  # Haut-droite
    quadrant_3 = image[milieu_y:hauteur, 0:milieu_x]  # Bas-gauche
    quadrant_4 = image[milieu_y:hauteur, milieu_x:largeur]  # Bas-droite
    
    print(f"✓ Image découpée en 4 quadrants de {milieu_x}x{milieu_y} pixels")
    
    # Sauvegarder les quadrants
    cv2.imwrite('images/chat_quadrant_1.jpg', quadrant_1)
    cv2.imwrite('images/chat_quadrant_2.jpg', quadrant_2)
    cv2.imwrite('images/chat_quadrant_3.jpg', quadrant_3)
    cv2.imwrite('images/chat_quadrant_4.jpg', quadrant_4)
    
    # Créer une mosaïque en réorganisant les quadrants
    haut = np.hstack((quadrant_2, quadrant_1))  # Inverser haut
    bas = np.hstack((quadrant_4, quadrant_3))  # Inverser bas
    mosaique = np.vstack((haut, bas))
    
    cv2.imwrite('images/chat_mosaique.jpg', mosaique)
    print("✓ Quadrants et mosaïque sauvegardés")
```

# ============================================================================
# CELLULE 11 : Filtres - Flou (Blur)
# ============================================================================

```python
"""
OpenCV offre plusieurs types de flou :
- Flou moyen (Average Blur)
- Flou gaussien (Gaussian Blur) - le plus utilisé
- Flou médian (Median Blur) - bon pour réduire le bruit
- Flou bilatéral (Bilateral Filter) - préserve les contours
"""

if image is not None:
    print("\n" + "="*60)
    print("FILTRES - FLOU")
    print("="*60)
    
    # Flou 1 : Flou moyen (Average Blur)
    kernel_size = (15, 15)
    image_flou_moyen = cv2.blur(image, kernel_size)
    print(f"✓ Flou moyen appliqué (kernel {kernel_size})")
    
    # Flou 2 : Flou gaussien (le plus populaire)
    image_flou_gaussien = cv2.GaussianBlur(image, (15, 15), 0)
    print("✓ Flou gaussien appliqué (15x15)")
    
    # Flou 3 : Flou médian (bon pour le bruit salt-and-pepper)
    image_flou_median = cv2.medianBlur(image, 15)
    print("✓ Flou médian appliqué (15)")
    
    # Flou 4 : Flou bilatéral (préserve les contours)
    image_flou_bilateral = cv2.bilateralFilter(image, 9, 75, 75)
    print("✓ Flou bilatéral appliqué")
    
    # Sauvegarder
    cv2.imwrite('images/chat_flou_moyen.jpg', image_flou_moyen)
    cv2.imwrite('images/chat_flou_gaussien.jpg', image_flou_gaussien)
    cv2.imwrite('images/chat_flou_median.jpg', image_flou_median)
    cv2.imwrite('images/chat_flou_bilateral.jpg', image_flou_bilateral)
    print("✓ Tous les flous sauvegardés")
```

# ============================================================================
# CELLULE 12 : Filtres supplémentaires (Netteté, Détection de contours)
# ============================================================================

```python
"""
Autres filtres utiles :
- Netteté (Sharpening)
- Détection de contours (Canny, Sobel)
- Dilatation et Érosion
"""

if image is not None:
    print("\n" + "="*60)
    print("FILTRES SUPPLÉMENTAIRES")
    print("="*60)
    
    # Filtre 1 : Netteté (Sharpening) avec un kernel personnalisé
    kernel_nettete = np.array([[-1, -1, -1],
                               [-1,  9, -1],
                               [-1, -1, -1]])
    image_nette = cv2.filter2D(image, -1, kernel_nettete)
    print("✓ Filtre de netteté appliqué")
    
    # Filtre 2 : Détection de contours avec Canny
    image_grise = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    image_contours = cv2.Canny(image_grise, 100, 200)
    print("✓ Détection de contours (Canny) appliquée")
    
    # Filtre 3 : Sobel (détection de contours avec gradient)
    sobel_x = cv2.Sobel(image_grise, cv2.CV_64F, 1, 0, ksize=5)
    sobel_y = cv2.Sobel(image_grise, cv2.CV_64F, 0, 1, ksize=5)
    sobel_combine = cv2.magnitude(sobel_x, sobel_y)
    sobel_combine = np.uint8(sobel_combine)
    print("✓ Détection de contours (Sobel) appliquée")
    
    # Sauvegarder
    cv2.imwrite('images/chat_nettete.jpg', image_nette)
    cv2.imwrite('images/chat_contours_canny.jpg', image_contours)
    cv2.imwrite('images/chat_contours_sobel.jpg', sobel_combine)
    print("✓ Filtres supplémentaires sauvegardés")
```

# ============================================================================
# CELLULE 13 : Résumé final de toutes les opérations
# ============================================================================

```python
"""
Affichage de toutes les opérations effectuées sur l'image.
"""

if image is not None:
    print("\n" + "="*60)
    print("RÉSUMÉ COMPLET DES OPÉRATIONS")
    print("="*60)
    print("✓ Téléchargement de l'image")
    print("✓ Chargement avec OpenCV")
    print("✓ Redimensionnement (3 méthodes)")
    print("✓ Crop (recadrage)")
    print("✓ Inversion (négatif)")
    print("✓ Rotation (90°, 180°, 45°)")
    print("✓ Découpage en quadrants + mosaïque")
    print("✓ Filtres de flou (moyen, gaussien, médian, bilatéral)")
    print("✓ Filtres supplémentaires (netteté, contours)")
    print("\n" + "="*60)
    print(f"📁 Toutes les images sont dans le dossier 'images/'")
    print("="*60)
```
