```python
import cv2
import matplotlib.pyplot as plt

# Lecture image
img = cv2.imread("image/chien.png")

if img is None:
    raise ValueError("Image non trouvée")

# 1️⃣ Flip horizontal (miroir gauche-droite)
flip_h = cv2.flip(img, 1)  # 1 → horizontal

# 2️⃣ Flip vertical (miroir haut-bas)
flip_v = cv2.flip(img, 0)  # 0 → vertical

# 3️⃣ Rotation selon un angle (centre, angle en degrés, échelle)
angle = 45
(h, w) = img.shape[:2]
center = (w//2, h//2)
scale = 1.0

M = cv2.getRotationMatrix2D(center, angle, scale)
rotated = cv2.warpAffine(img, M, (w, h))

# Affichage
plt.figure(figsize=(12,6))

plt.subplot(2,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original")
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(cv2.cvtColor(flip_h, cv2.COLOR_BGR2RGB))
plt.title("Flip horizontal")
plt.axis('off')

plt.subplot(2,2,3)
plt.imshow(cv2.cvtColor(flip_v, cv2.COLOR_BGR2RGB))
plt.title("Flip vertical")
plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.title(f"Rotation {angle}°")
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Lecture image
img = cv2.imread("image/chien.png")

if img is None:
    raise ValueError("Image non trouvée")

# Taille du patch
patch_h, patch_w = 64, 64

# Pas de la fenêtre
stride_h, stride_w = 64, 64  # pas = taille → pas de recouvrement

# Liste pour stocker les patchs
patches = []

# Dimensions de l'image
H, W = img.shape[:2]

# Sliding window
for y in range(0, H - patch_h + 1, stride_h):
    for x in range(0, W - patch_w + 1, stride_w):
        patch = img[y:y+patch_h, x:x+patch_w]
        patches.append(patch)

print(f"Nombre de patchs : {len(patches)}")

# Affichage des 4 premiers patchs
plt.figure(figsize=(10,4))
for i in range(min(4, len(patches))):
    plt.subplot(1,4,i+1)
    plt.imshow(cv2.cvtColor(patches[i], cv2.COLOR_BGR2RGB))
    plt.title(f"Patch {i+1}")
    plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Lecture image
img = cv2.imread("image/chateau.png", cv2.IMREAD_GRAYSCALE)

if img is None:
    raise ValueError("Image non trouvée")

# Sobel gradient X (dérivée horizontale)
sobelx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)

# Sobel gradient Y (dérivée verticale)
sobely = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)

# Amplitude du gradient
gradient_magnitude = np.sqrt(sobelx**2 + sobely**2)
gradient_magnitude = np.uint8(np.clip(gradient_magnitude, 0, 255))  # Conversion pour affichage

# Affichage
plt.figure(figsize=(12,6))

plt.subplot(2,2,1)
plt.imshow(img, cmap='gray')
plt.title("Image originale")
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(np.abs(sobelx), cmap='gray')
plt.title("Gradient Sobel X")
plt.axis('off')

plt.subplot(2,2,3)
plt.imshow(np.abs(sobely), cmap='gray')
plt.title("Gradient Sobel Y")
plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(gradient_magnitude, cmap='gray')
plt.title("Amplitude du gradient")
plt.axis('off')

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Lecture image en niveaux de gris
img = cv2.imread("image/chateau.png", cv2.IMREAD_GRAYSCALE)

if img is None:
    raise ValueError("Image non trouvée")

# Application du filtre Laplacien
laplacian = cv2.Laplacian(img, cv2.CV_64F, ksize=3)  # ksize = taille du noyau

# Conversion pour affichage
laplacian_display = cv2.convertScaleAbs(laplacian)

# Affichage
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Image originale")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(laplacian_display, cmap='gray')
plt.title("Filtre Laplacien")
plt.axis('off')

plt.show()
```

```python
import requests
import os

def download_image(url: str, savefilename: str, timeout: int = 30):
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3",
        "Accept": "image/avif,image/webp,image/apng,image/*,*/*;q=0.8",
        "Accept-Language": "en-US,en;q=0.9",
    }

    r = requests.get(url, headers=headers, stream=True, timeout=timeout)

    try:
        r.raise_for_status()
    except requests.HTTPError as e:
        if r.status_code == 403:
            raise RuntimeError(
                "HTTP 403 (Forbidden) : le serveur bloque le téléchargement direct.\n"
                "- Essaie d'adapter les headers.\n"
                "- Si lien Google Drive / Dropbox → lien direct.\n"
                "- Certains sites bloquent le hotlink."
            ) from e
        raise

    with open(savefilename, 'wb') as f:
        for chunk in r.iter_content(chunk_size=8192):
            if chunk:
                f.write(chunk)

    return os.path.abspath(savefilename)


# Exemple d'utilisation
url = "https://upload.wikimedia.org/wikipedia/commons/6/6e/Golde33443.jpg"
path = download_image(url, "downloaded_image.jpg")
print(f"Image téléchargée et enregistrée sous : {path}")
```

```python
# charger l'image avec OpenCV
image= cv2.imread("downloaded_image.jpg", cv2.IMREAD_UNCHANGED)

#Veriifer si l'image a été chargée correctement
if image is None :
    print("Erreur lors du chargement de l'image.")
else :
    print("Image chargée avec succès.")
    #Afficher les dimensions de l'image
    print("Dimensions de l'image :", image.shape)
```

```python
# Afficher l'image avec opencv dans une fenetre externe opencv

#Verifier que l'image est bien chargée
if image is not None:
    cv2.imshow("Image", image)
    cv2.waitKey(0)  # Attendre une touche pour fermer la fenêtre
    cv2.destroyAllWindows()
else:
    print("Impossible d'afficher l'image, elle n'a pas été chargée correctement.")
```

```python
#je veux afficher un flou, filtre 1 1 1 1
kernel = np.array([[1, 0, 1],
                   [1, 0, 1],
                   [1, 0, 1]], dtype=np.float32) / 9.0
#Appliquer le filtre de flou à l'image
blurred_image = cv2.filter2D(image, -1, kernel)
#Afficher l'image floutée
if blurred_image is not None:
    cv2.imshow("Image Floutée", blurred_image)
    cv2.waitKey(0)  # Attendre une touche pour fermer la fenêtre
    cv2.destroyAllWindows()
else:
    print("Impossible d'afficher l'image floutée, elle n'a pas été traitée correctement.")
```

```python
#Je veux afficher les 2 images côte à côte
if image is not None and blurred_image is not None:
    combined_image = np.hstack((image, blurred_image))
    cv2.imshow("Image Originale et Floutée", combined_image)
    cv2.waitKey(0)  # Attendre une touche pour fermer la fenêtre
    cv2.destroyAllWindows()
```
